# Coder: a coding model for Claude Code

**Daily use:** press **Start Coder** (GitHub Actions). It runs INSTEAD of Jarvis (same address, same free GPU hours), so close Jarvis first. Ready in about 5-8 minutes; the log shows the address and the model's speed. It closes by itself after 20 minutes without a request.

**Before the first run** (see README, section 'Coder helper'): settings = GPU T4 x2 and Internet on; inputs = the `assistant-code` dataset; secrets ticked for this notebook (including `CODER_KEY`).

```python
import os, sys, zipfile
from pathlib import Path

# Filled in by the GitHub "Start Coder" button: a one-time key that opens this run's
# encrypted secrets (Kaggle gives no Kaggle Secrets to runs started that way).
# Leave it empty; starting from this editor uses your Kaggle Secrets as usual.
RUN_KEY = ""
if RUN_KEY:
    os.environ["PVA_RUN_KEY"] = RUN_KEY


def find_code():
    for p in Path("/kaggle/input").glob("**/coder_main.py"):  # zip unpacked by Kaggle
        if (p.parent / "main.py").exists():
            return p.parent
    for z in Path("/kaggle/input").glob("**/assistant-code*.zip"):  # zip kept as a file
        zipfile.ZipFile(z).extractall("/tmp/pva-code")
        return Path("/tmp/pva-code/src")
    return None


src = find_code()
assert src, "Attach your 'assistant-code' dataset: Add Input -> Your Work -> Datasets (see README)."
sys.path.insert(0, str(src))

import coder_main
coder_main.run()
```
