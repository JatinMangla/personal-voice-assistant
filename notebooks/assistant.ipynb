# Jarvis: personal voice assistant

**Daily use:** press your **Start Jarvis** button (GitHub), or here: **Save Version → Save & Run All (Commit)** and stop the editor session. Jarvis is ready in about 4–5 minutes at your ngrok address.

It closes by itself after 10 minutes without use (memory is saved first), which also frees the GPU.

**Before the first run** (see README): settings = GPU T4 x2 and Internet on; inputs = `assistant-code` dataset and the `build-model-cache` notebook output; secrets ticked for this notebook.

```python
import sys, zipfile
from pathlib import Path


def find_code():
    for p in Path("/kaggle/input").glob("**/app.py"):  # zip unpacked by Kaggle
        if (p.parent / "main.py").exists():
            return p.parent
    for z in Path("/kaggle/input").glob("**/assistant-code*.zip"):  # zip kept as a file
        zipfile.ZipFile(z).extractall("/tmp/pva-code")
        return Path("/tmp/pva-code/src")
    return None


src = find_code()
assert src, "Attach your 'assistant-code' dataset: Add Input -> Your Work -> Datasets (see README)."
sys.path.insert(0, str(src))

import main
main.run()
```
